```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\Jyoti Hanagandi\OneDrive\Desktop\cyber_threat_detection\data\malicious_phish.csv"
)

print(f"Columns: {df.columns.tolist()}")
print(f"Shape: {df.shape}")
print(f"\nLabel counts:\n{df['type'].value_counts()}")

df.head(10)
```

```python

import os
import pandas as pd

df["label"] = df["type"].apply(lambda x: 0 if x == "benign" else 1)

print(f"Total URLs: {len(df)}")
print(f"Safe (0):      {len(df[df['label'] == 0])}")
print(f"Malicious (1): {len(df[df['label'] == 1])}")

df = df.drop_duplicates(subset=["url"])
df = df.dropna(subset=["url"])
df = df.reset_index(drop=True)

print(f"\nAfter cleaning: {len(df)}")

data_path = r"C:\Users\Jyoti Hanagandi\OneDrive\Desktop\cyber_threat_detection\data"
os.makedirs(data_path, exist_ok=True)

output_path = os.path.join(data_path, "raw_urls.csv")
df[["url", "label"]].to_csv(output_path, index=False)

print("STEP 1 COMPLETE!")
print("Saved to:", output_path)
```
